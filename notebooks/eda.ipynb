```python
df_imports = pd.read_parquet(f'{files_path}/fish_trade_import_200601_202603.parquet')
df_exports = pd.read_parquet(f'{files_path}/fish_trade_export_200601_202603.parquet')
```

```python
df_imports = df_imports.astype({
    'metricFOB': float,
    'metricKG': float,
    'year': int,
    'monthNumber': int,
})
```

```python
display(df_imports.info())
```

```python
print('Mostrando o top 10 estados que mais importaram em relação ao valor de importação no ano de 2025')
display(
    df_imports[df_imports['year'] == 2025]
    .groupby(['state'])['metricFOB']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
```

```python
intervals = [
    ('2006-01', '2010-12'),
    ('2011-01', '2015-12'),
    ('2016-01', '2020-12'),
    ('2021-01', '2023-12'),
    ('2024-01', '2026-03'),
]

for from_date, to_date in intervals:
    payload_test = {
        'flow': 'export',
        'monthDetail': True,
        'period': {'from': from_date, 'to': to_date},
        'filters': [{'filter': 'heading', 'values': ['0303']}],
        'details': ['country', 'state', 'city', 'heading'],
        'metrics': ['metricFOB', 'metricKG'],
    }

    resp = requests.post('https://api-comexstat.mdic.gov.br/cities', json=payload_test, params={'language': 'pt'})
    data = resp.json()['data']['list']
    df = pd.DataFrame(data)
    
    print(f'\n--- {from_date} até {to_date} ---')
    print(f'Total registros: {len(data)}')
    print(f'Meses únicos: {sorted(df["monthNumber"].unique())}')
    print(f'Meses por ano:')
    print(df.groupby("year")["monthNumber"].nunique())
    print(resp.status_code)
    print(resp.json())
    
    sleep(10)
```
