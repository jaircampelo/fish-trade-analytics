```python
#===========================================================#
#                       Importing libs                      #
#===========================================================#
import requests
import pandas as pd
from typing import Literal
from pathlib import Path
from datetime import date
from urllib.parse import quote_plus
from dotenv import load_dotenv
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus
from time import sleep
import logging
import os
import re
```

```python
files_path = Path(os.getcwd()).resolve().parent / 'data' / 'raw'
print(files_path)
```

```python
df_imports = pd.read_parquet(f'{files_path}/fish_trade_import_200601_202603.parquet')
df_exports = pd.read_parquet(f'{files_path}/fish_trade_export_200601_202603.parquet')
```

```python
display(df_imports.info())
```

```python
df_imports = df_imports.astype({
    'metricFOB': float,
    'metricKG': float,
    'year': int,
    'monthNumber': int,
})
```

```python
display(df_imports.info())
```

```python
print('Mostrando o top 10 estados que mais importaram em relação ao valor de importação no ano de 2025')
display(
    df_imports[df_imports['year'] == 2025]
    .groupby(['state'])['metricFOB']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
```

```python
display(df_exports.info())
```

```python
df_exports = df_exports.astype({
    'metricFOB': float,
    'metricKG': float,
    'year': int,
    'monthNumber': int,
})
```

```python
display(df_exports.info())
```

```python
print('Mostrando o top 10 estados que mais exportaram em relação ao valor de importação no ano de 2025')
display(
    df_exports[df_exports['year'] == 2025]
    .groupby(['state'])['metricFOB']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
```
